```python
# DBSCAN

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_moons

X,y = make_moons(n_samples=1000, noise=0.08)
plt.scatter(X[:,0], X[:, 1], c=y, cmap='cividis', edgecolors='black', linewidths=0.3)
```

```python
#Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

plt.scatter(X_scaled[:,0], X_scaled[:, 1], c=y, cmap='cividis', edgecolors='black', linewidths=0.3)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps = 0.3)
dbscan.fit(X_scaled)
```

```python
dbscan.labels_
```

```python
plt.scatter(X_scaled[:,0], X_scaled[:, 1], c=dbscan.labels_, cmap='cividis', edgecolors='black', linewidths=0.3)
```
