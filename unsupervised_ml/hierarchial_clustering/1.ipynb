```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Import IRIS Dataset

from sklearn.datasets import load_iris

iris = load_iris()

iris_data = pd.DataFrame(iris.data)
```

```python
iris_data.columns = iris.feature_names
```

```python
iris_data
```

```python
iris.target
```

```python
#Standard Scaling

from sklearn.preprocessing import StandardScaler

scalar = StandardScaler()

X_scaled = scalar.fit_transform(iris_data)
```

```python
X_scaled
```

```python
## Applying the PCA

from sklearn.decomposition import PCA

pca = PCA(n_components=2)
```

```python
pca_scaled = pca.fit_transform(X_scaled)
```

```python
plt.scatter(pca_scaled[:,0], pca_scaled[:,1], edgecolors='black', linewidths=0.5, c=iris.target)
```

```python
## Aglomerative Clustering

# To construct a Dendogram

import scipy.cluster.hierarchy as sc

plt.figure(figsize=(15,7))
plt.title("Dendogram")
sc.dendrogram(sc.linkage(pca_scaled, method = 'ward'))
plt.xlabel("Sample Index")
plt.ylabel("Euclidian Distance")
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster = AgglomerativeClustering(n_clusters=2, linkage='ward')
cluster.fit(pca_scaled)
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0], pca_scaled[:,1], edgecolors='black', linewidths=0.5, c=cluster.labels_)
```
